```python
#1. Find the length of a string. 
string="Hello, world!"
string_length=len(string)
print(string_length)
```

```python
#2. Reverse a string. 
string="HELLO"
reversed_string=string[::-1]
print(reversed_string)
```

```python
#3. Check whether a string is a palindrome. 
string="MADAM"
reverse_string=string[::-1]
if string==reverse_string:
    print("it is palindrome")
else:
    print("it is not a palindrome")
```

```python
#4. Count vowels and consonants. 
string="pythonprogramming"
vowels_count=0
consonants_count=0
for char in string:
    if char.lower() in "aeiou":
        vowels_count+=1
    elif char.isalpha():
        consonants_count+=1
print("vowels:",vowels_count)
print("consonants:",consonants_count)
```

```python
#5. Count uppercase and lowercase characters.
string="Hello World And Python"
print("Uppercase letters are:")
for char in string:
    if char>='A' and char<='Z':
        print(char,end=" ")
print()
print()
print("Lowercase letters are:")
for char in string:
    if char>='a' and char<='z':
        print(char,end=" ")
```

```python
#6. Count digits and spaces in a string. 
string = "Hello @123 hi 143"
count_digit=0
count_space=0
print("digits count=")
for char in string:
    if char>='0' and char<='9':
        count_digit+=1
print(count_digit)
print()
print()
print("Spaces count=")
for char in string:
    if char==' ':
        count_space+=1
print(count_space)
```

```python
#7. Count the frequency of each character. 
string = "python programming"
frequency ={}
for char in string:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1
for key, value in frequency.items():
    print(f"'{key}': {value}")
```

```python
#8. Find the first non-repeating character.
string = "swiss"
frequency = {}

# Step 1: Count the frequency of each character
for char in string:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

# Step 2: Find the first character with a frequency of 1
for char in string:
    if frequency[char] == 1:
        print("First non-repeating character:", char)
        break # We found the first one, so we stop the loop
```

```python
#9. Remove duplicate characters from a string.
string = "programming"
result = ""

for char in string:
    # If the character is not already in our result string, add it
    if char not in result:
        result += char

print("String without duplicates:", result)
```

```python
#10. Check whether two strings are anagrams.
string1 = "listen"
string2 = "silent"

# It's good practice to make them lowercase and remove spaces just in case
str1_clean = string1.replace(" ", "").lower()
str2_clean = string2.replace(" ", "").lower()

# Sort the characters and compare them
if sorted(str1_clean) == sorted(str2_clean):
    print(f"'{string1}' and '{string2}' are anagrams!")
else:
    print(f"'{string1}' and '{string2}' are NOT anagrams.")
```
