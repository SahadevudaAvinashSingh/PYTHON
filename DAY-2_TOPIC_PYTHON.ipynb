```python
#1. Print numbers from 1 to N. 
N = int(input("Enter how many numbers to print:"))
for i in range(1, N+1):
    print(i,end=" ")
```

```python
#2. Print numbers from N to 1. 
N = int(input("Enter how many numbers to print:"))
for i in range(1,N+1):
    print(N,end=" ")
    N=N-1
```

```python
#3. Print all even numbers from 1 to N. 
N = int(input("Enter number:"))
for i in range(1, N+1):
    if i%2==0:
        print(i,end=" ")
```

```python
#4. Print all odd numbers from 1 to N. 
N = int(input("Enter number:"))
for i in range(1, N+1):
    if i%2==1:
        print(i,end=" ")
```

```python
#5. Find the sum of numbers from 1 to N. 
N = int(input("Enter number:"))
total=0
for i in range(1, N+1):
    total+=i
print(total)
```

```python
#6. Find the factorial of a number. 
n = int(input("Enter number to get factorial:"))
fac=1
for i in range(1, n+1):
    fac*=i
print(fac)
```

```python
#.7 Print the multiplication table of a number. 
n=int(input("Enter number to get the table:"))
i=1
while i<=10:
    print(f"{n} x {i} = {n*i}")
    i=i+1
```

```python
#8. Count the number of digits in an integer. 
n=int(input("enter number to get count:"))
count=0
if n==0:
    count=1

while n>0:
    count+=1
    n=n//10
    
    
print("count= ",count)
```

```python
#9. Find the sum of digits of a number. 
n=int(input("enter number to get sum:"))
add=0
n=abs(n)
while n>0:
    digit=n%10
    add+=digit
    n=n//10
print("sum of digits of a number is :",add)
```

```python
#10. Reverse a number. 
n = int(input("Enter number to reverse:"))
print("Reverse of number is :")
while n>0:
    digit=n%10
    print(digit,end="")
    n=n//10
```
