```python
#1. Even or Odd
#Take an integer and determine whether it is even or odd.
n = int(input("Enter number to check even or odd:"))
if n%2==0:
    print("Even number")
else:
    print("Odd number")
```

```python
#2. Positive, Negative, or Zero
#Given a number, determine whether it is positive, negative, or zero.
n = int(input("Enter number to check:"))
if n>0:
    print("Positive")
elif n<0:
    print("Negative")
else:
    print("Zero")
```

```python
#3. Largest of Two Numbers
#Find the larger of two numbers without using max().
a = int(input("enter a value:"))
b = int(input("enter b value:"))
if a>b:
    print(f"'a'={a} is largest")
else:
    print(f"'b'={b} is largest")
```

```python
#4. Largest of Three Numbers
#Find the largest among three numbers without using max().
a = int(input("enter a value:"))
b = int(input("enter b value:"))
c = int(input("enter c value:"))
if a>b and a>c:
    print(f"'a' = {a} is largest")
elif b>c:
    print(f"'b' = {b} is largest")
else:
    print(f"'c' = {c} is largest")
```

```python
#5. Smallest of Three Numbers
#Find the smallest among three numbers without using min().
a = int(input("enter a value:"))
b = int(input("enter b value:"))
c = int(input("enter c value:"))
if a<b and a<c:
    print(f"'a' = {a} is smallest")
elif b<c:
    print(f"'b' = {b} is smallest")
else:
    print(f"'c' = {c} is smallest")
```

```python
#6. Divisible by 5 and 11
#Check whether a number is divisible by both 5 and 11.
n = int(input("enter number to check:"))
if n%5==0 and n%11==0:
    print(f"{n} is divisible by both 5 and 11.")
else:
    print(f"{n} is not divisible by both 5 and 11.")
```

```python
#7. Leap Year
#Determine whether a given year is a leap year.
y = int(input("enter year to check leap or not:"))
if (y%4==0 and y%100!=0) or (y%400==0) :
    print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
#9. Check whether a character is a vowel or consonant. 
ch = input("Enter character to Check whether a character is a vowel or consonant:")
if (ch=='a' or ch=='e' or ch=='i' or ch=='o' or ch=='u' or ch=='A' or ch=='E' or ch=='I' or ch=='O' or ch=='U'):
    print("given character is a VOWEL")
else:
    print("given character is a CONSONANT")
```

```python
#10. Check whether a person is eligible to vote based on age. 
age = int(input("Enter age to check eligibility:"))
if age>=18:
    print(f"age = {age} : you are eligible to cast vote")
else:
    print(f"age = {age} : you are not eligible to cast vote because your age is less than '18'")
```

```python
#11. Calculate a student's grade from their marks. 
m = float(input("Enter marks to get grades :"))
if m>=90:
    print("A+ grade")
elif m>=80 and m<90:
    print("A grade")
elif m>=70 and m<80:
    print("B grade")
elif m>=60 and m<70:
    print("C grade")
elif m>=50 and m<60:
    print("D grade")
elif m>=40 and m<50:
    print("E grade")
else:
    print("FAIL")
```

```python
#12. Build a simple calculator using if/elif/else. 
a = float(input("Enter a value:"))
op = input("Enter operator:")
b = float(input("Enter b value:"))
if op=='+':
    print(f"ADD = {a+b}")
elif op=='-':
    print(f"SUB = {a-b}")
elif op=='*':
    print(f"MUL = {a*b}")
elif op=='/':
    print(f"DIV = {a/b}")
```
