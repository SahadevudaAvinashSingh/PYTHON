#### RESTAURANT BILL WITH DISCOUNT

```python
# -------------------------------
# RESTAURANT BILLING SYSTEM
# -------------------------------

menu = {
    "Burger": 150,
    "Pizza": 350,
    "Sandwich": 120,
    "French Fries": 100,
    "Coke": 50,
    "Coffee": 80
}
# above menu is in dict data type
# below is functions
def display_menu():
    print("\n========== MENU ==========")
    for item, price in menu.items():           #this is loops concept
        print(f"{item:<15} ₹{price}")
    print("==========================\n")


def take_order():
    orders = []              #this is List data type

    while True:              # this is loops concept
        item = input("Enter item name: ").title()

        if item not in menu:          # this is conditional statements concept
            print("Item not found in menu!")
            continue

        try:                                   #this is exceptional handling concept(try, except)
            quantity = int(input("Enter quantity: "))

            if quantity <= 0:
                print("Quantity must be greater than 0.")
                continue

        except ValueError:
            print("Please enter a valid number.")
            continue

        orders.append((item, quantity))

        choice = input("Do you want to order more? (yes/no): ").lower()

        if choice != "yes":
            break

    return orders


def calculate_subtotal(orders):
    subtotal = 0

    for item, qty in orders:
        subtotal += menu[item] * qty

    return subtotal


def calculate_discount(subtotal):
    if subtotal >= 2000:
        return 15
    elif subtotal >= 1000:
        return 10
    elif subtotal >= 500:
        return 5
    else:
        return 0


def print_bill(orders):

    subtotal = calculate_subtotal(orders)

    discount_percent = calculate_discount(subtotal)

    discount = subtotal * discount_percent / 100

    after_discount = subtotal - discount

    gst = after_discount * 0.05

    final_bill = after_discount + gst

    print("\n")
    print("=" * 40)
    print("        RESTAURANT BILL")
    print("=" * 40)

    for item, qty in orders:
        price = menu[item]
        total = price * qty

        print(f"{item:<15}{qty} x ₹{price:<5}= ₹{total}")

    print("-" * 40)

    print(f"Subtotal           : ₹{subtotal:.2f}")
    print(f"Discount ({discount_percent}%)       : -₹{discount:.2f}")
    print(f"After Discount     : ₹{after_discount:.2f}")
    print(f"GST (5%)           : ₹{gst:.2f}")

    print("-" * 40)
    print(f"TOTAL BILL         : ₹{final_bill:.2f}")
    print("=" * 40)

    print("Thank You!")
    print("Visit Again")


# ---------------- MAIN PROGRAM ----------------

display_menu()

orders = take_order()

print_bill(orders)
```

#### NETFLIX SUBSCRIPTION PLAN SELECTOR

```python
# NETFLIX SUBSCRIPTION PLAN SELECTOR

# 1. Gather User Input
budget = float(input("Enter your monthly budget(₹): "))
users = int(input("Enter number of people will use the account?: "))
tv = input("Do you watch on TV? [yes/no]: ").lower()
quality = input("What Quality do you prefer? [HD / FULLHD / 4K]: ").upper()
student = input("Are you a student? [yes/no]: ").lower()

# 2. Determine Plan and Pricing Logic
recommended_plan = "None"
monthly_cost = 0
reason = "Budget too low or requirements too high."

if quality == "4K" or users > 2 or (tv == "yes" and budget >= 650):
    if budget >= 650:
        recommended_plan = "Premium"
        monthly_cost = 650
        reason = "Best option for multiple users with TV and 4K support."
elif quality == "FULLHD" or users == 2 or (tv == "yes" and budget >= 500):
    if budget >= 500:
        recommended_plan = "Standard"
        monthly_cost = 500
        reason = "Great for couples or 1080p viewing on a TV."
elif quality == "HD" or tv == "yes" or budget >= 250:
    if budget >= 250:
        recommended_plan = "Basic"
        monthly_cost = 250
        reason = "Good for standard viewing on a TV or tablet."
elif budget >= 150:
    recommended_plan = "Mobile"
    monthly_cost = 150
    reason = "Most affordable option for a single mobile device."

# 3. Calculate Yearly Cost (apply student discount if applicable)
yearly_cost = monthly_cost * 12
if student == "yes":
    # Applying a hypothetical 10% student discount on yearly plans
    yearly_cost = int(yearly_cost * 0.90) 

# 4. Formatted Output
print("\n----------- Netflix Plan Selector -----------")
print(f"{'Budget':<19}: ₹{int(budget)}")
print(f"{'Users':<19}: {users}")
print(f"{'Watch on TV':<19}: {tv.capitalize()}")
print(f"{'Need 4K':<19}: {'Yes' if quality == '4K' else 'No'}")
print(f"{'Student':<19}: {student.capitalize()}")
print(f"{'Recommended Plan':<19}: {recommended_plan}")
print(f"{'Monthly Cost':<19}: ₹{monthly_cost}")
print(f"{'Yearly Cost':<19}: ₹{yearly_cost}")
print(f"{'Reason':<19}: {reason}")
print("\nEnjoy your streaming! 🍿")
```
