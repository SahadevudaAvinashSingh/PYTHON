```python
#1. Print a square pattern using *. 
for i in range(4):
    for j in range(4):
        print("*",end=" ")
    print()
```

```python
#2. Print a right-angled triangle pattern. 
for i in range(4):
    for j in range(i+1):
        print("*",end=" ")
    print()
```

```python
#3. Print an inverted triangle pattern. 
for i in range(4):
    for j in range(j+1):
        print("*",end=" ")
        j-=1
    print()
```

```python
#4. Print a pyramid pattern. 
n=4
for i in range(n):
    for j in range(n-i-1):             #it will print spaces 
        print("",end=" ")
    for k in range(i+1):               #it will print stars
        print("*",end=" ")
    print()
```

```python
#5. Print a number triangle
for i in range(4):
    for j in range(i+1):
        print(i+1,end=" ")
    print()
```

```python
#6. Write a function to check whether a number is even. 
def check_evenodd(number):
    if number%2==0:
        print(f"{number} is an even number.")
    else:
        print(f"{number} is an odd number.")

a=int(input("Enter a value:"))
b=int(input("Enter b value:"))
check_evenodd(a)
check_evenodd(b)
check_evenodd(10)
```

```python
#7. Write a function to check whether a number is prime. 
def check_prime(num):
    if num<1:
        print("Not a prime number")
    else:
        for i in range(2,num):
            if num%i==0:
                print("Not a prime number")
                break
        else:
            print("It is a prime number")

number=int(input("Enter value to check prime or not:"))
check_prime(number)
```

```python
#8. Write a function to calculate factorial. 
def factorial(num):
    fact=1
    for i in range(1,num+1):
        fact=fact*i
    print(fact)
number=int(input("Enter value to calculate the factorial:"))
factorial(number)
```

```python
#9. Write a function to calculate the sum of digits. 
def add(num):
    add=0
    for i in range(1,num+1):
        x=num%10
        add=add+x
        num=num//10
    print(add)
number = int(input("enter value:"))
add(number)
```
