```python
#1. Check whether a number is prime.
n = int(input("enter number to check prime or not:"))
if n<=1:                          #here 0 and 1 are not prime number so we write this condition
    print("Not a Prime number")
else:
    for i in range(2,int(n**0.5)+1):            #here if n=7 then int(n**0.5)+1 = int(root(7))+1 = int(2.64)+1 = 2+1=3  so range(2,3) so the i=2 only
        if n%i==0:                              #so i=2 then n%i==0 => 7%2==1 so it is a prime number
            print("Not a Prime number")
            break
    else:
        print("It is a Prime number")
```

```python
# 2. Print all prime numbers between 1 and N.
n = int(input("enter number to check all prime numbers between:"))
for num in range(2,n+1):
    is_prime= True
    for i in range(2,int(num**0.5)+1):
        if num%i==0:
            is_prime=False
            break
    if is_prime:
        print(num)
```

```python
#3. Count the number of prime numbers between 1 and N. 
n = int(input("enter number to check no of prime numbers:"))
count=0
for num in range(2,n+1):
    is_prime= True
    for i in range(2,int(num**0.5)+1):
        if num%i==0:
            is_prime=False
            break
    if is_prime:
        count+=1
print(count)
```

```python
#4. Check whether a number is a palindrome.
n = int(input("enter number to check palindrome or not:"))
num=n
rev=0
while num>0:
    digit=num%10
    rev=(rev*10)+digit
    num=num//10
if rev==n:
    print(" palindrome")
else:
    print("not a palindrome")
```

```python
#6. Check whether a number is an Armstrong number. 
n = int(input("Enter number to check Armstrong number or not:"))
power = len(str(n))
num=n
arm=0
while num>0:
    x=num%10
    arm=(arm)+(x**power)
    num=num//10
if arm==n:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
#7. Find the GCD of two numbers. 
#Method 1: Using the Euclidean Algorithm (Without imports)
num1=int(input("Enter 1st value:"))
num2=int(input("Enter 2nd value:"))
a=num1
b=num2
while b!=0:
    temp=b
    b=a%b
    a=temp
print("the GCD of the two numbers is :",a)

#Method 2: Using the built-in math module
import math
num1=int(input("Enter 1st value:"))
num2=int(input("Enter 2nd value:"))
gcd=math.gcd(num1,num2)
print("the GCD of the two numbers is :",gcd)
```

```python
#9. Print the Fibonacci series up to N terms. 
n = int(input("Enter the number of terms (N): "))
n1=0
n2=1
count=0

if n<=0:
    print("Please enter a positive integer higher than 0.")
elif n==1:
    print("Fibonacci sequence up to 1 term:")
    print(n1)
else:
    print("Fibonacci sequence:")
    while count<n:
        print(n1,end=" ")

        nth=n1+n2
        n1=n2
        n2=nth

        count+=1
```
